```python
# Section-A :Data Structures & Control Statements
# Create a list of 5 integers
numbers = [10, 20, 30, 40, 50]
print("Original List:", numbers)

# Append
numbers.append(60)
print("After append:", numbers)

# Extend
numbers.extend([70, 80])
print("After extend:", numbers)

# Insert
numbers.insert(2, 25)
print("After insert:", numbers)

# Remove
numbers.remove(40)
print("After remove:", numbers)

# Pop
numbers.pop()
print("After pop:", numbers)

# Index
print("Index of 30:", numbers.index(30))

# Count
print("Count of 20:", numbers.count(20))

# Sort
numbers.sort()
print("After sort:", numbers)

# Reverse
numbers.reverse()
print("After reverse:", numbers)

# Clear
numbers.clear()
print("After clear:", numbers)
```

```python
# Create a tuple
students = ("Ravi", "Sita", "Rahul")

print(students)

# Try to change the second name
students[1] = "Priya"

print(students)
```

# Explanation 
A tuple is immutable, so we cannot change its elements after it is created. Therefore, Python displays a TypeError.

```python
# Create a set with duplicate values
numbers = {10, 20, 30, 20, 40, 10, 50}
print(numbers)
```

# Explanation
A set stores only unique values. It automatically removes duplicate elements. So, 10 and 20 appear only once in the output. Also, sets do not maintain the insertion order, so the elements may be displayed in a different order.

```python
# Check positive, even, zero, or negative

num = 8

if num > 0:
    if num % 2 == 0:
        print("Positive and Even")
    else:
        print("Positive and Odd")
elif num == 0:
    print("Zero")
else:
    print("Negative")
```

```python
# Section C: Exploring Pandas
import pandas as pd

# Create a DataFrame
data = {
    "Name": ["Ravi", "Sita", "Ram", "Anu", "Kiran", "Ravi", "Priya", "Raj", "Rani", "Ravi"],
    "Age": [20, 21, 22, 23, 24, 20, 25, 26, 27, 20],
    "City": ["Kadapa", "Hyderabad", "Chennai", "Delhi", "Mumbai", "Kadapa", "Pune", "Goa", "Vizag", "Kadapa"],
    "Salary": [25000, 30000, None, 35000, 40000, 25000, 45000, None, 50000, 25000]
}

df = pd.DataFrame(data)

# Display information
print(df.info())

# Display statistics
print(df.describe())

# Select Name and City columns
print(df[["Name", "City"]])

# Drop City column
df = df.drop("City", axis=1)
print(df)

# Fill missing Salary values with mean
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print(df)

# Remove duplicate rows
df = df.drop_duplicates()
print(df)
```
