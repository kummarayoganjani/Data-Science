```python
#Checking Missing Values:
df.isnull().sum()
```

```python
#Fill Missing Values:
df=df.fillna(df.mean(numeric_only=True))
```

```python
df
```

```python
df.dtypes
```

```python
#Detect Outliers:
import seaborn as sns
import matplotlib.pyplot as plt
#Checking Boxplots
for i in df.select_dtypes(include=['int','float']).columns:
    sns.boxplot(df[i])
    plt.show()
```

```python
#Outlier Capping
for i in df.select_dtypes(include=['int','float']).columns:
    q1=df[i].quantile(0.25)
    q3=df[i].quantile(0.75)
    IQR=q3-q1
    lower=q1-1.5*IQR
    upper=q3+1.5*IQR
    df[i]=df[i].clip(lower,upper)
```

```python
#Checking outliers after capping
for col in df.select_dtypes(include=['int','float']).columns:
    sns.boxplot(df[col])
    plt.title(f'Box Plot of {col}')
    plt.show()
```

```python
# 2.Statistical Methods:
```

```python
df.describe()
```

```python
#Mean & Median:
mean_values = df.mean(numeric_only=True) 
median_values = df.median(numeric_only=True)
print('Mean')
print(mean_values)
print('\nMedian')
print(median_values)
```

```python
#Standard Deviation & Inter Quartile Range:
std_values = df.std(numeric_only=True)

q1 = df.quantile(0.25, numeric_only=True)
q3 = df.quantile(0.75, numeric_only=True)
iqr = q3 - q1

print('Standard Deviation')
print(std_values)

print('\nIQR')
print(iqr)
```

 #####  Interesting findings in Statistical Summary:

 The statistical summary shows that most numerical values are concentrated around their average values. The LB variable has a moderate spread, indicating relatively stable fetal heart rate measurements. The AC and FM variables show greater variation, which suggests differences in fetal activity among observations. Some variables have a large range between minimum and maximum values, indicating possible outliers. The difference between mean and median in a few variables suggests that the data may be slightly skewed.

```python
#3.Data Visualization:
```

```python
#Histogram:
cols = ['LB', 'AC', 'FM', 'UC']

df[cols].hist(figsize=(8,6), bins=20)
plt.suptitle('Histograms of Numerical Variables')
plt.show()
```

```python
#Boxplot:
plt.figure(figsize=(8,5))
sns.boxplot(data=df[['LB', 'AC', 'FM', 'UC']])
plt.title('Boxplots of Numerical Variables')
plt.show()
```

```python
#Barchart:
df['NSP'].value_counts().plot(kind='bar')

plt.title('Frequency of NSP Categories')
plt.xlabel('NSP')
plt.ylabel('Count')
plt.show()
```

```python
#Pie Chart:
df['NSP'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Percentage of NSP Categories')
plt.ylabel('')
plt.show()
```

```python
#Correlation HeatMap:
plt.figure(figsize=(10,8))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, cmap='coolwarm')
plt.title('Correlation Heatmap') 
plt.show()
```

```python
#Scatterplot:
sns.scatterplot(x='LB', y='AC', data=df)
plt.title('LB vs AC') 
plt.show()
```

```python
#Pairplot:
sns.pairplot(df[['LB','AC','FM','UC']]) 
plt.show()
```

```python
#Violin plot:
sns.violinplot(y=df['LB']) 
plt.title('Violin Plot of LB')
plt.show()
```
