```python
#Problem Statement : Hospital Patient Data Analysis
import pandas as pd
#Load patient dataset:
df = pd.read_csv("Patient_Data.csv")
df
```

```python
print(df.head())
df.info()
```

```python
billing_columns = df[['PatientID', 'Department', 'Doctor', 'BillAmount']]

print(billing_columns)
```

```python
df = df.drop(['ReceptionistID', 'CheckInTime'], axis=1)

print(df)
```

```python
total_bill = df.groupby('Department')['BillAmount'].sum()

print(total_bill)
```

```python
df = df.drop_duplicates(subset='PatientID')

print(df)
```

```python
mean_bill = df['BillAmount'].mean()

df['BillAmount'] = df['BillAmount'].fillna(mean_bill)

print(df)
```

```python
#Loading the Billing Dataset:
billing_df = pd.read_csv("Billing_Data.csv")
billing_df
```

```python
#Merging the two datasets:
merged_df = pd.merge(df, billing_df, on='PatientID')

print(merged_df)
```

```python
#Adding new patient rows:
new_patients = pd.DataFrame({
    'PatientID': [106, 107],
    'Name': ['John', 'Sara'],
    'Department': ['Cardiology', 'Neurology'],
    'Doctor': ['Dr. Smith', 'Dr. Kumar'],
    'BillAmount': [4000, 5000]
})

df = pd.concat([df, new_patients], ignore_index=True)

print(df)
```

```python
#Adding new billing columns:
billing_categories = billing_df[['InsuranceCovered', 'FinalAmount']]

final_df = pd.concat([df, billing_categories], axis=1)

print(final_df)
```
