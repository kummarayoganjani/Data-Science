## Summary of Dataset Key Features:

            The SONAR (Mines vs. Rocks) dataset contains 208 samples. Each sample has 60 numerical features representing the energy of the sonar signal in different frequency bands. The target variable has 2 classes: Mine (M) and Rock (R). Among the 208 samples, 111 belong to the Mine class and 97 belong to the Rock class. Therefore, this dataset is a binary classification dataset used for underwater object detection.

```python
####Data preprocessing steps including data normalization:
from sklearn.preprocessing import MinMaxScaler
# Separate features and target
x = df.iloc[:, :-1]
y = df.iloc[:, -1]
# Normalize features
scaler = MinMaxScaler()
x_scaled = scaler.fit_transform(x)
x_scaled[:5]
```

## Data Preprocessing:

    * Missing values were checked during data exploration, and no missing values were found; therefore, no treatment was required.

    * Duplicate records were checked, and no duplicate rows were found in the dataset.

    * The feature values were normalized using MinMaxScaler.

## Conclusion:

        The SONAR dataset did not contain any missing values or duplicate records. Therefore, only feature normalization was performed, scaling all input values to the range 0–1 for better machine learning and deep learning model performance.

```python
####2.Model Implementation:
from sklearn.preprocessing import LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
# Encode target variable (M and R)
encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)
# Build ANN model
model = Sequential()
# Hidden layer
model.add(Dense(16, activation='relu', input_shape=(60,)))
# Output layer
model.add(Dense(1, activation='sigmoid'))
# Compile model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
# Display model summary
model.summary()
```

## ANN Model Implementation:

The ANN model was implemented using the TensorFlow/Keras deep learning library. The model consists of an input layer with 60 features, one hidden layer with 16 neurons using the ReLU activation function, and an output layer with one neuron using the Sigmoid activation function. Since the problem is a binary classification task involving Mine and Rock classes, the Sigmoid activation function is suitable for the output layer. The model contains a total of 993 trainable parameters, including weights and biases. The model was compiled using the Adam optimizer and binary cross-entropy loss function.


```python
#####Dataset into training and test sets:
from sklearn.model_selection import train_test_split
##Splitting the dataset:
x_train, x_test, y_train, y_test = train_test_split(
    x_scaled,
    y_encoded,
    test_size=0.2,
    random_state=42,
    stratify=y_encoded
)
##Checking the training and testing sizes:
print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)
##Checking target sizes:
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

The dataset was divided into training and testing sets using an 80:20 ratio. The training set contains 80% of the data for training the ANN model, while the remaining 20% is used to evaluate the model on unseen data. Stratified splitting was used to maintain a similar distribution of Mine and Rock classes in both sets.


```python
#####Model Training and Prediction:
# Train the ANN model
history = model.fit(
    x_train,
    y_train,
    epochs=50,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
# Predict on test data
y_pred_probability = model.predict(x_test)
# Convert probabilities to class labels
y_pred = (y_pred_probability >= 0.5).astype(int)
# Display first 10 predictions
print("Predicted values:")
print(y_pred[:10].flatten())
print("Actual values:")
print(y_test[:10])
```

## Model Training and Prediction:

The ANN model was trained using the training dataset for 50 epochs with a batch size of 16. A validation split of 20% was used to monitor the model during training. After training, the model was used to predict the classes of the test samples. The sigmoid output probabilities were converted into binary class labels using a threshold of 0.5.

```python
import pandas as pd
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

import keras_tuner as kt
```

```python
def build_model(hp):

    model = Sequential()

    neurons = hp.Choice(
        'neurons',
        values=[16, 32, 64]
    )

    activation = hp.Choice(
        'activation',
        values=['relu', 'tanh']
    )

    hidden_layers = hp.Choice(
        'hidden_layers',
        values=[1, 2]
    )

    model.add(
        Dense(
            neurons,
            activation=activation,
            input_shape=(60,)
        )
    )

    if hidden_layers == 2:
        model.add(
            Dense(16, activation=activation)
        )

    model.add(
        Dense(1, activation='sigmoid')
    )

    learning_rate = hp.Choice(
        'learning_rate',
        values=[0.001, 0.0001]
    )

    model.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
tuner = kt.RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=6,
    directory='sonar_new_search',
    project_name='ann_tuning_new'
)
```

```python
tuner.search(
    x_train,
    y_train,
    epochs=20,
    batch_size=16,
    validation_split=0.2,
    verbose=1
)
```

```python
###Get best hyperparameters:
best_hps = tuner.get_best_hyperparameters(
    num_trials=1
)[0]
print("Best Number of Neurons:",
      best_hps.get('neurons'))
print("Best Activation Function:",
      best_hps.get('activation'))
print("Best Number of Hidden Layers:",
      best_hps.get('hidden_layers'))
print("Best Learning Rate:",
      best_hps.get('learning_rate'))
```

```python
###Get best model:
best_model = tuner.get_best_models(
    num_models=1
)[0]
```

```python
###Evaluate best model:
best_loss, best_accuracy = best_model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("Best Random Search Model Test Accuracy:",
      best_accuracy)
```

```python
###Show all Random Search results:
tuner.results_summary()
```

## Explanation:
Random Search was adopted as a structured approach for hyperparameter tuning. Six different combinations of neurons, activation functions, hidden layers, and learning rates were tested. Validation accuracy was used to select the best configuration. The best configuration achieved a validation accuracy of 82.35% and was then evaluated on the test dataset to measure its performance on unseen data.

```python
####4.Evaluation:
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
# Make predictions on test data
y_pred_prob = best_model.predict(x_test, verbose=0)
# Convert probabilities to class labels
y_pred = (y_pred_prob >= 0.5).astype(int).ravel()
# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
# Display results
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-Score:", f1)
```
