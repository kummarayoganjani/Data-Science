```python
#Outlier capping
for i in df.select_dtypes(include=['int','float']).columns:
    q1=df[i].quantile(0.25)
    q3=df[i].quantile(0.75)
    IQR=q3-q1
    lower_bound=q1-1.5*IQR
    upper_bound=q3+1.5*IQR
    df[i]=df[i].clip(lower_bound,upper_bound)
```

```python
#Checking Outliers again after capping
for i in df.select_dtypes(include=['int','float']).columns:
    sns.boxplot(df[i])
    plt.show()
```

```python
###Task 2: Dimensionality Reduction with PCA
x = df.drop('Type', axis=1)
y = df['Type']
```

```python
###1.Standardize the Features:
###Standardization:
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
x_scaled[:5]
```

## Explanation:
 The numerical features were standardized using StandardScaler. Standardization transformed each feature to have a mean of 0 and a standard deviation of 1. This step is important before applying PCA because PCA is sensitive to the scale of the variables.

```python
###2.Determine Optimal Number of Principal Components:
###Applying PCA to all features:
from sklearn.decomposition import PCA
pca = PCA()
pca.fit(x_scaled)
```

```python
from sklearn.decomposition import PCA
# Apply PCA
pca = PCA(n_components=2)
# Transform the standardized data
x_pca = pca.fit_transform(x_scaled)
# Show first 5 rows
x_pca[:5]
```

```python
###Scree Plot:
plt.plot(range(1, len(pca.explained_variance_ratio_) + 1), pca.explained_variance_ratio_, marker='o')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.title('Scree Plot')
plt.show()
```

```python
####Cumulative Explained Variance:
cum_var = pca.explained_variance_ratio_.cumsum() 
cum_var
```

## Explanation:
        The optimal number of principal components was determined using the scree plot and cumulative explained variance. The scree plot shows how much variance is explained by each component, and the cumulative explained variance shows the total variance explained as components are added. The number of components was selected where the curve begins to flatten and where a high percentage (around 90–95%) of the total variance is retained.

```python
###3.Transform Dataset into Principal Components:
##Select number of components
pca_final = PCA(n_components=2)
pca_final
```

```python
###Transform the data:
x_pca = pca_final.fit_transform(x_scaled)
x_pca
```

```python
###Create a new DataFrame:
pca_df = pd.DataFrame(x_pca, columns=['PC1', 'PC2'])
pca_df.head()
```

## Explanation:
    The original standardized dataset was transformed into principal components using PCA. The new variables PC1 and PC2 represent the reduced-dimensional form of the dataset while retaining most of the important information from the original features.

```python
####Task 3: Clustering with Original Data
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
```

```python
###1.Applying K-Means Clustering:
##Create K-Means model
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans
```

```python
###Fit the model and get cluster labels
clusters = kmeans.fit_predict(x)
clusters
```

```python
####Add cluster labels to the dataset
df['Cluster'] = clusters
df[['Cluster']].head()
```

## Explanation:
    K-Means clustering was applied to the original dataset with three clusters. The model assigned each observation to a cluster, and the resulting cluster labels were added to the dataset for further analysis and visualization.

```python
###2.Visualizing Clustering Results:
###Scatter plot:
plt.figure(figsize=(8,6))
plt.scatter(x.iloc[:,0], x.iloc[:,1], c=clusters)
plt.xlabel(x.columns[0])
plt.ylabel(x.columns[1])
plt.title('K-Means Clustering on Original Data')
plt.show()
```

## Explanation:
        A scatter plot was used to visualize the K-Means clustering results based on the first two features of the original dataset. Different colors represent different clusters, which helps in observing the separation and grouping of data points.

```python
####3.Evaluate Clustering Performance:
###Silhouette Score
sil_score = silhouette_score(x, clusters)
print('Silhouette Score:', sil_score)
```

```python
####Davies–Bouldin Index
db_score = davies_bouldin_score(x, clusters)
print('Davies-Bouldin Index:', db_score)
```

## Explanation:
        The clustering performance was evaluated using the Silhouette Score and the Davies–Bouldin Index. A higher Silhouette Score indicates better cluster quality, while a lower Davies–Bouldin Index indicates better separation between clusters.

```python
###Task 4: Clustering with PCA Data
####1.Apply K-Means to PCA-transformed data
from sklearn.cluster import KMeans
# Create K-Means model
kmeans_pca = KMeans(n_clusters=3, random_state=42)
# Fit the model on PCA data 
clusters_pca = kmeans_pca.fit_predict(x_pca) # Display first 5 cluster labels 
clusters_pca[:5]
```

## Explanation:
        K-Means clustering was applied to the PCA-transformed dataset using three clusters. The model assigned a cluster label to each observation in the PCA data, and the first five cluster labels were displayed.

```python
####2.Visualize clustering results:
import matplotlib.pyplot as plt 
plt.figure(figsize=(8,6))
plt.scatter(x_pca[:,0], x_pca[:,1], c=clusters_pca)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('K-Means Clustering on PCA Data') 
plt.show()
```

## Explanation:
        A scatter plot was created using the first two principal components (PC1 and PC2) to visualize the clustering results obtained from the PCA-transformed data. Different colors represent different clusters, which helps in observing the separation of clusters in reduced-dimensional space.

```python
####3.Compare clustering results:
from sklearn.metrics import silhouette_score, davies_bouldin_score
# Original data scores 
sil_original = silhouette_score(x, clusters) 
db_original = davies_bouldin_score(x, clusters) 
# PCA data scores 
sil_pca = silhouette_score(x_pca, clusters_pca)
db_pca = davies_bouldin_score(x_pca, clusters_pca)
print('Original Data Silhouette Score:', sil_original)
print('PCA Data Silhouette Score:', sil_pca)
print('Original Data Davies-Bouldin Index:', db_original)
print('PCA Data Davies-Bouldin Index:', db_pca)
```

## Explanation:
    The clustering results obtained from the PCA-transformed data were compared with those from the original dataset. PCA reduced the dimensionality of the data while preserving most of the important information. The PCA-based clustering showed clearer cluster separation and simpler visualization, while the original dataset contained more features and higher dimensionality.

```python
###Task 5: Comparison and Analysis
comparison = pd.DataFrame({ 'Dataset': ['Original Data', 'PCA Data'], 'Silhouette Score': [sil_original, sil_pca], 'Davies-Bouldin Index': [db_original, db_pca] }) 
comparison
```

## 1.Comparing the clustering results:

        The clustering results of the original dataset and PCA-transformed dataset were compared using Silhouette Score and Davies–Bouldin Index. The scores indicate the quality of clustering and the separation between clusters.

## 2.Similarities and differences:

        Both the original dataset and the PCA-transformed dataset produced similar cluster groups. However, the PCA-transformed data showed clearer separation between clusters, while the original dataset had a more complex distribution due to the larger number of features.

## 3.Impact of dimensionality reduction:
        Dimensionality reduction using PCA simplified the dataset by reducing the number of features while retaining most of the important information. This improved cluster visualization and reduced computational complexity.

## 4.Trade-offs between PCA and original data:

        Clustering on the original dataset uses all available feature information but may include noise and redundant variables. PCA reduces dimensionality and speeds up computation, but some information may be lost during transformation. Therefore, PCA offers simpler analysis, while the original dataset preserves complete information.

## 2.Practical implications of PCA and clustering:

Clustering helps discover hidden patterns and group similar observations without predefined labels. Together, PCA and clustering can improve data interpretation, visualization, and computational efficiency in practical data analysis tasks.

## 3.Recommendations:

        Use PCA when the dataset contains many correlated features or when dimensionality reduction and visualization are important. Use clustering directly on the original dataset when preserving all feature information is necessary. Use PCA before clustering when the dataset is high-dimensional, contains redundant variables, or requires faster computation and clearer cluster visualization.
