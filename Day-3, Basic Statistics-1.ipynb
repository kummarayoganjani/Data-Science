```python
#Descriptive Analytics and Data Preprocessing on Sales & Discounts Dataset
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Loading the dataset
df = pd.read_csv("sales_data_with_discounts.csv")
df
```

```python
print(df.head())
print(df.info())
```

```python
#Identifying numerical columns in the dataset.
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

print(numerical_columns)
```

```python
#Mean, Median, Mode, and Standard deviation.
for column in numerical_columns:
    print("\nColumn:", column)
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode()[0])
    print("Standard Deviation:", df[column].std())
```

# Brief interpretation of these statistics:
The mean shows the average value of the data. The median represents the middle value, and the mode shows the most frequently occurring value. The standard deviation shows how much the data values vary from the mean.Overall, the statistics help us understand the average, central values, and variation in the sales data.

```python
#Data Visualization:
#Histograms:
for column in numerical_columns:
    plt.hist(df[column], bins=10)
    plt.title("Histogram of " + column)
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.show()
```

```python
#Boxplots:
for column in numerical_columns:
    plt.boxplot(df[column])
    plt.title("Box Plot of " + column)
    plt.ylabel(column)
    plt.show()
```

```python
#Identifying categorical columns in the dataset.
categorical_columns = df.select_dtypes(include=['object']).columns

print(categorical_columns)
```

```python
#Bar Charts
for column in categorical_columns:
    df[column].value_counts().plot(kind='bar')
    plt.title("Bar Chart of " + column)
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.show()
```

# CONCLUSION:
The descriptive analysis helped to understand the distribution and variation of numerical variables in the sales dataset. Mean, median, mode, and standard deviation were calculated for all numerical columns. Histograms showed that several sales-related variables were skewed and contained high values. Box plots helped identify possible outliers. Bar charts were used to analyze categorical variables and understand the frequency of different categories. Overall, the analysis provides useful insights into sales, discounts, and product-related data.
