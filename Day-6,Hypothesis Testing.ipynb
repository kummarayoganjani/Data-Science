```python
###  a.Build 99% Confidence Interval Using Sample Standard Deviation:#####
```

```python
import numpy as np
from scipy import stats
```

```python
data = [1.13, 1.55, 1.43, 0.92, 1.25,
        1.36, 1.32, 0.85, 1.07, 1.48,
        1.20, 1.33, 1.18, 1.22, 1.29]
```

```python
#Finding Sample Size:
n = len(data)
print("Sample size:", n)
```

```python
#Calculating Sample Mean:
mean = np.mean(data)
print("Sample mean:", round(mean, 4))
```

```python
#Calculating Sample Standard Deviation:
s = np.std(data, ddof=1)
print("Sample standard deviation:", round(s, 4))
```

```python
#Finding t Critical Value:
confidence = 0.99
alpha = 1 - confidence

t_value = stats.t.ppf(1 - alpha/2, df=n-1)
print("t critical value:", round(t_value, 4))
```

```python
#Calculating Margin of Error:
margin_error_t = t_value * (s / np.sqrt(n))
print("Margin of error:", round(margin_error_t, 4))
```

```python
#Calculating 99% confidence interval:
lower_t = mean - margin_error_t
upper_t = mean + margin_error_t

print("99% CI using sample standard deviation:")
print("Lower limit:", round(lower_t, 4))
print("Upper limit:", round(upper_t, 4))
```

###### Steps:
I entered the 15 durability values into Python and calculated the sample mean, sample standard deviation, and sample size. Using a 99% confidence level, I found the t-critical value and calculated the confidence interval as mean ± margin of error. The t-distribution was used because the population standard deviation is unknown and the sample size is small (15). Therefore, the t-distribution provides a more reliable estimate of the true mean durability of the print-heads.


```python
##### b.Build 99% Confidence Interval Using Known Population Standard Deviation:#######
```

```python
##Set Population Standard Deviation:
##Given ,sigma=0.2
sigma=0.2
```

```python
#Finding z-critical value:
z_value = stats.norm.ppf(1 - alpha/2)
print("z critical value:", round(z_value, 4))
```

```python
##Calculating margin of error:
margin_error_z = z_value * (sigma / np.sqrt(n))
print("Margin of error:", round(margin_error_z, 4))
```

```python
##Calculating 99% confidence interval:
lower_z = mean - margin_error_z 
upper_z = mean + margin_error_z 
print("99% CI using known population standard deviation:") 
print("Lower limit:", round(lower_z, 4))
print("Upper limit:", round(upper_z, 4))
```
