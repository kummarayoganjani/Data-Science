```python
# Exercise 1: Prime Numbers
num = int(input("Enter a number: "))

if num <= 1:
    print(num, "is not a Prime number")
else:
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print(num, "is a Prime number")
    else:
        print(num, "is not a Prime number")
```

```python
# Exercise 2: Product of Random Numbers
import random

a = random.randint(1, 20)
b = random.randint(1, 20)

print("Multiply:", a, "*", b)
answer = int(input("Enter your answer: "))

if answer == a * b:
    print("Correct Answer!")
else:
    print("Wrong Answer!")
    
```

```python
# Exercise 3: Squares of Even/Odd Numbers
# Program to print squares of even numbers from 100 to 200

for i in range(100, 201):
    if i % 2 == 0:   # Choosing even numbers
        print(i, "=", i * i)
```

```python
#Exercise 4:
# Problem Statement (Question): Hospital Billing System

room_charge = 2000
doctor_fee = 1500
lab_test_charge = 300

days = 4
tests = 3
medicine_charge = 2400

room_total = room_charge * days
lab_total = lab_test_charge * tests

total_bill = room_total + doctor_fee + lab_total + medicine_charge

if total_bill > 10000:
    discount = total_bill * 0.10
else:
    discount = 0

final_bill = total_bill - discount

print("Room Charges =", room_total)
print("Doctor Fee =", doctor_fee)
print("Lab Test Charges =", lab_total)
print("Medicine Charges =", medicine_charge)
print("Total Bill =", total_bill)
print("Discount =", discount)
print("Final Bill =", final_bill)
```
