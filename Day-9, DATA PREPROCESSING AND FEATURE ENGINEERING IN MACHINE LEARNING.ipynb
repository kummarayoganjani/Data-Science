```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        df[col].fillna(df[col].mean(), inplace=True)

print(df.isnull().sum())
```

```python
###Separate Numerical Columns:
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
```

```python
num_cols
```

```python
#####Applying Scaling Techniques to Numerical Features:
##A.Standard Scaling:
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df_standard = df.copy()
df_standard[num_cols] = scaler.fit_transform(df_standard[num_cols])
```

```python
df_standard[num_cols]
```

```python
df_standard[num_cols].head()
```

```python
####B.Min-Max Scaling:
from sklearn.preprocessing import MinMaxScaler

mmscaler = MinMaxScaler()
df_minmax = df.copy()
df_minmax[num_cols] = mmscaler.fit_transform(df_minmax[num_cols])
```

```python
df_minmax[num_cols]
```

```python
df_minmax[num_cols].head()
```

## The scenarios where each scaling technique is preferred and why:
   
   Standard Scaling:
     * Best for algorithms like Logistic Regression, SVM, KNN, PCA.
     * Works well when data is approximately normally distributed.
   
   
   
   Min-Max Scaling:
     * Best for Neural Networks and distance-based methods when a fixed range is preferred.
     * Sensitive to outliers.

```python
####2.Encoding Techniques:
######●	Apply One-Hot Encoding to categorical variables less than 5:
cat_cols = df.select_dtypes(include=['object']).columns
```

```python
cat_cols
```

```python
small_cat = [col for col in cat_cols if df[col].nunique() < 5]
small_cat
```

```python
df_onehot = pd.get_dummies(df, columns=small_cat,drop_first=True)
df_onehot
```

```python
df_onehot.head()
```

```python
####● Use Label Encoding for categorical variables:(more than 5 categories)
from sklearn.preprocessing import LabelEncoder

df_label = df.copy()
le = LabelEncoder()
large_cat = [col for col in cat_cols if df[col].nunique() >= 5]
for col in large_cat:
    df_label[col] = le.fit_transform(df_label[col])
```

```python
(df_label[large_cat].head())
```

##  pros and cons of One-Hot Encoding and Label Encoding:
  
 #### One-Hot Encoding:
         Pros:
            No ordinal relationship is created.
            Good for low-cardinality categorical variables.
         Cons:
            Creates many new columns.
 #### Label Encoding:
         Pros:
             Simple and memory efficient.
             Useful for tree-based models.
         Cons:
             May create a false order among categories.

```python
#####3.Feature Engineering:
 ####Creating 2 new Features:
```

```python
# Feature 1: Age group
df['age_group'] = pd.cut(df['age'],
                         bins=[0, 25, 45, 65, 100],
                         labels=['Young', 'Adult', 'Middle', 'Senior'])
df[['age', 'age_group']]
```

```python
df[['age', 'age_group']].head()
```

```python
# Feature 2: Hours category
df['work_hours_type'] = np.where(df['hours_per_week'] >= 40,
                                 'Full-time',
                                 'Part-time')
df[['hours_per_week','work_hours_type']]
```

```python
df[['hours_per_week','work_hours_type']].head()
```

## Rationale of choosing Two New Features:

        Two new categorical features were created to improve the predictive power of the model. age_group represents different life stages, while work_hours_type represents employment intensity (full-time or part-time). These engineered features help the model identify meaningful patterns related to income and may improve classification performance.

```python
####●Applying a transformation to atleast one skewed numerical feature:
df['capital_gain_log'] = np.log1p(df['capital_gain'])
```

```python
df[['capital_gain', 'capital_gain_log']]
```

```python
df[['capital_gain', 'capital_gain_log']].head()
```
